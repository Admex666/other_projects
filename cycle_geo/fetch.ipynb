```python
import pandas as pd
from services import slope_and_elevation, population, temperature

bike_usage = pd.read_csv('bike_usage_2019.csv')
```

```python
for i, row in bike_usage.iterrows():
    print(f"{i}/{len(bike_usage)}")
    print(row['City'])
    lat, lon, average_slope, elevation = slope_and_elevation(row['City'], row['Country'])
    pop = population(row['Country'])
    temp_q1, temp_median, temp_q3 = temperature(lat, lon)
    print("=========================")
    items = [lat, lon, average_slope, elevation, pop, temp_q1, temp_median, temp_q3]
    bike_usage.loc[i, ['lat', 'lon', 'average_slope', 
                    'elevation', 'pop', 'temp_q1', 
                    'temp_median', 'temp_q3']] = items
    
    
```

```python
# Save
bike_usage.to_csv('bike_usage_extended.csv', index=False)
```
