```python
# Lekérdezés
import os
import json
import requests

# 1️⃣ SerpApi kulcs (helyettesítsd a saját kulcsoddal)
SERPAPI_API_KEY = os.getenv("FLIGHTAPI_KEY")

# 2️⃣ Keresési paraméterek
params = {
    "engine": "google_flights",
    "api_key": SERPAPI_API_KEY,
    "departure_id": "JFK",
    "arrival_id": "DFW",
    "type": 2,
    "outbound_date": "2025-09-30"
}


# 3️⃣ Lekérdezés
search = requests.get("https://serpapi.com/search", params=params)
response = search.json()
print(json.dumps(response, indent=2))
```

```python
keys = ["best_flights", "other_flights"]
for key in keys:
    flights = response.get(key, [])

    if not flights:
        print("Nincs találat a megadott dátumokra.")
    else:
        for item in flights:
            flight = item['flights']
            price = item.get('price', 9999)
            if (type(flight) == list) and (price != 9999):
                departure = flight[0].get('departure_airport', "N/A").get('time', '1970-01-01 00:00')
                arrival = flight[0].get('arrival_airport', "N/A").get('time', '1970-01-01 00:00')
                airline = flight[0].get('airline', "N/A")

                print(f"{airline}: {departure} -> {arrival} | Ár: {price}")
```
